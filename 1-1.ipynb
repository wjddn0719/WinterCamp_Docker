```python
import numpy as np

words = ["apple", "banana", "dog", "cat"]
vector = [[0.9, 0.2], [0.8, 0.3], [0.1, 0.9], [0.2, 0.8]]
```

```python
def cos_sin(A, B):
    return np.dot(A, B) / (np.linalg.norm(A) * np.linalg.norm(B))
```

```python
print(cos_sin(vector[2], vector[3]))
```

```python
# 단어 -> 인덱스 매핑
word_to_index = {}
index = 0
```

```python
for w in words:
    word_to_index[w] = index
    index = index + 1
    print(w, ":", word_to_index[w])
```

```python
def one_hot(word):
    V = len(words) # 단어 사전 크기
    vec = np.zeros(V) # [0, 0, 0, 0] 으로 초기화
    if word in word_to_index:
        idx = word_to_index[word]
        vec[idx] = 1.0
    return vec
```

```python
for w in words:
    print(w, "→", one_hot(w))
```

```python
# 단어별 위치를 딕셔너리로 저장
word_to_idx = {word: idx for idx, word in enumerate(words)}

# 3. 임베딩 초기화 : 정규분포(평균=0, 표준편차=0.1), 임베딩 행렬의 크기(4x4)
#              적당한 범위(≈ -0.3 ~ +0.3) 로 제한
W_embed = np.random.normal(0, 0.1, size=(len(words), 4))

# 4. 랜덤모델 적용
def get_embedding(word):
    idx = word_to_idx.get(word)
    if idx is not None:
        return W_embed[idx]
    else:
        return None
        
print("=== 임베딩 결과 ===")
for w in words:
    print(w, "→", get_embedding(w))
```

```python
import tiktoken

def split_text(full_text, chunksize):
    encoder = tiktoken.encoding_for_model("gpt-5")
    total_encoding = encoder.encode(full_text)
    # print(len(total_encoding))

    total_token_count = len(total_encoding)
    text_list = []
    for i in range(0, total_token_count, chunksize):
        chunk = total_encoding[i: i+chunksize]
        decoded = encoder.decode(chunk)
        text_list.append(decoded)
    return text_list
```

```python
split_text(txt, 70)
```

```python
# 랜덤 모델을 사용한 문장 임베딩(단어 -> 벡터)

# 같은 값을 넣으면 동일한 벡터 값이 나오도록 수정!

def random_token_embeddings(texts, dim = 786):

    token_vecs=[]
    for token in texts:
        # 토큰 문자열 해시 → 시드
        token_seed = abs(hash(token)) % (2**32)
        # 같은 문장이라면 동일한 벡터값으로, 아니면 걍 랜덤
        rng = np.random.default_rng(token_seed)
        # 우리 벡터는 768 차원
        token_vec = rng.normal(0, 0.1, size=(dim,))
        token_vecs.append(token_vec)
    return np.vstack(token_vecs)
```

```python
random_token_embeddings(txt)
```
